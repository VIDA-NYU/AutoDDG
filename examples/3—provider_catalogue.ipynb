# Explore AutoDDG supported LLMs providers

Explore providers, metadata, and models exposed by AutoDDG via `LiteLLM` — [litellm.ai](https://www.litellm.ai)

```python
from autoddg import AutoDDG

factory = AutoDDG()
```

## Let's list out providers we support and have tested-out!

`list_providers()` shows the canonical provider identifiers bundled with AutoDDG.


```python
factory.list_providers()
```

## Inspect a provider parameters passed to LiteLLM

For beginners, no need to care much here, but for more advanced users, more can be passed through here in cascade to LiteLLM. E.g. aliases, base URL overrides, expected environment variables, etc.

```python
factory.describe_provider('openai')
```

## Discover available models per providers now

`list_model_names()` returns the LiteLLM-catalogued models for the provider.

📣 WARNING: Please recall that LiteLLM is a broad library, hence here what's of interest to you are chat-based LLMs, you can therefore ignore vision-based models and so on.


```python
factory.list_model_names('mistral')
```

# TheFuzz W/ AutoDDG, no errors anymore!
